## Numpy array and basics

```python
import numpy as np 
```

### Creating array from list


```python
arr_1d = np.array([1,2,3,4,5])
print("1D array:", arr_1d)

arr_2d = np.array([[1,2,3], [4,5,6]])
print("2D array:", arr_2d)
```

### List vs numpy array


```python
py_list= [1,2,3,4,5] # vector multiplication
print("python list multiplication:", py_list*2)

np_array = np.array([1,2,3]) # element wise multiplication
print("numpy array multiplication:", np_array*2)

# let's test the time efficiency of numpy array multiplication vs python list multiplication
import time 
start = time.time()
py_list = [i * 2 for i in range(100000000)]
print("\n List operation time: ", time.time() - start)


start = time.time()
np_array = np.arange(100000000) * 2
print("\n numpy operation time: ", time.time() - start)

```

### Creating array from Scratch


```python
zeros = np.zeros((3,4))
print("zeros array: \n", zeros)

ones= np.ones((3,4))
print("ones array: \n",ones )

full = np.full((2,2), 7)        # np.full(shape, full_value)
print("full array: \n", full)

# random = np.random.random((3,4))
random = np.random.randint(1,10, size=(3,4)) # np.random.randint(low, high, size)
print("random array: \n",random)

sequence = np.arange(0,10,3) # start, stop, step
print("sequence array: \n", sequence)
```

### Vector, Matrices and Tensor

```python
vector = np.array([1,2,3])
print("vector:",vector)


matrix = np.array([[1,2,3],
                   [1,2,5],
                   [1,5,9]])
print("matrix: \n", matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor: \n", tensor)
```

### Array properties


```python
arr = np.array([[1,2,3],
                [4,5,6],])
print("shape of arr:",arr.shape)
print("Dimention of arr:", arr.ndim)
print("size of arr:", arr.size)
print("data type of arr: ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("original array",arr)

reshaped = arr.reshape(3,4)
print("\n Reshaped array:", reshaped)

# flatten(returns copy of array, instead of view)
flattened = reshaped.flatten()
print("\n Flattened array:", flattened)

# Ravel(returns view, instead of copy)
raveled = reshaped.ravel()
print("\n Raveled array:", raveled)

# Transpose
transpose = reshaped.T
print("\n Transpose of array:", transpose)
```
